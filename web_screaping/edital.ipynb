```python
import pandas as pd
import re
from io import StringIO

def criar_tabela_do_edital(caminho_arquivo: str) -> pd.DataFrame:
    """
    Lê um arquivo de texto de um edital, extrai os dados dos candidatos
    usando expressões regulares e os retorna em um DataFrame do pandas.

    Args:
        caminho_arquivo: O caminho para o arquivo .txt contendo os dados.

    Returns:
        Um pandas DataFrame com as informações tabuladas e tipadas.
    """
    try:
        with open(caminho_arquivo, 'r', encoding='utf-8') as f:
            conteudo = f.read()
    except FileNotFoundError:
        print(f"Erro: O arquivo '{caminho_arquivo}' não foi encontrado.")
        return pd.DataFrame() # Retorna um DataFrame vazio em caso de erro

    # Expressão regular para capturar os 7 campos de cada candidato.
    # Esta expressão é robusta o suficiente para lidar com espaçamentos variados.
    # Padrão: (inscrição), (nome), (nota1), (acertos1), (nota2), (acertos2), (nota_final)
    padrao = re.compile(r"(\d+),\s*(.*?),\s*([\d\.]+),\s*(\d+),\s*([\d\.]+),\s*(\d+),\s*([\d\.]+)")

    # Encontra todas as correspondências no texto.
    # re.findall retorna uma lista de tuplas, onde cada tupla contém os grupos capturados.
    dados_extraidos = padrao.findall(conteudo)

    # Define os nomes das colunas para o DataFrame
    colunas = [
        "Inscrição", "Nome", "Nota_Conhecimentos_Básicos", "Acertos_Básicos",
        "Nota_Conhecimentos_Específicos", "Acertos_Específicos", "Nota_Final"
    ]

    # Cria o DataFrame a partir dos dados extraídos
    df = pd.DataFrame(dados_extraidos, columns=colunas)

    # Converte as colunas para os tipos de dados apropriados para futuras análises
    tipos_de_dados = {
        "Inscrição": int,
        "Nome": str,
        "Nota_Conhecimentos_Básicos": str,
        "Acertos_Básicos": int,
        "Nota_Conhecimentos_Específicos": str,
        "Acertos_Específicos": int,
        "Nota_Final": str
    }
    df = df.astype(tipos_de_dados)

    return df, conteudo

# --- Bloco de Execução Principal ---
if __name__ == "__main__":
    # 1. Simulação: Criando o arquivo 'edital.txt' com os seus dados.
    # Em um caso real, este arquivo já existiria no seu computador.
    conteudo_txt = "10088194, Abilio da Cruz Ramos Neto, 32.00, 41, 36.00, 53, 68.00 / 10075745, Abimaeli Simplicio Vasques, 38.00, 41, 48.00, 56, 86.00 / 10136648, Abraao Flavio Nascimento Souza, 21.00, 24, 29.00, 38, 50.00 / 10013311, Abraao Lincoln de Sousa Alves, 24.00, 37, 30.00, 50, 54.00 / 10000737, Abraao Linconies da Silva Campelo, 16.00, 33, 24.00, 47, 40.00 / 10103946, Abraao Prudencio Viana Costa, 29.00, 36, 36.00, 50, 65.00 / 10110884, Acassi Batista Flores, 22.00, 35, 33.00, 48, 55.00 / 10001306, Acauane Ohana Severino Vaz, 23.00, 30, 24.00, 39, 47.00 / 10008570, Adaias Abner Brito Silva, 31.00, 37, 44.00, 55, 75.00 / 10096873, Adailson Lourenco Salvino, 25.00, 35, 37.00, 51, 62.00 / 10130030, Adailton Rios da Silva, 28.00, 38, 37.00, 50, 65.00 / 10093650, Adalberto Sennhauser, 18.00, 34, 42.00, 56, 60.00 / 10124975, Adalcyr Luiz da Silva Junior, 26.00, 38, 36.00, 53, 62.00 / 10118053, Adam Alves Borges, 24.00, 37, 43.00, 56, 67.00 / 10067120, Adamo Henrique Louzada, 25.00, 35, 43.00, 53, 68.00 / 10102807, Adao Gomes Santana, 15.00, 27, 31.00, 41, 46.00 / 10064585, Adayanna Karolline de Lima Moreira Schneider, 17.00, 33, 23.00, 38, 40.00"
    nome_do_arquivo = "data/edital_text.txt"
    # with open(nome_do_arquivo, "w", encoding="utf-8") as f:
    #     f.write(conteudo_txt)

    # 2. Processamento: Chamar a função para ler e tabular os dados.
    tabela_edital, conteudo = criar_tabela_do_edital(nome_do_arquivo)

    # 3. Exibição: Imprimir o DataFrame resultante.
    # Usar to_string() para garantir que todas as linhas sejam mostradas.
    if not tabela_edital.empty:
        print("Dados do Edital Tabulados com Sucesso:")
        print(tabela_edital.to_string())


```

```python
import pandas as pd
import re
from weasyprint import HTML, CSS

# (Esta é a mesma função da resposta anterior)
def criar_tabela_do_edital(caminho_arquivo: str) -> pd.DataFrame:
    try:
        with open(caminho_arquivo, 'r', encoding='utf-8') as f:
            conteudo = f.read()
    except FileNotFoundError:
        print(f"Erro: O arquivo '{caminho_arquivo}' não foi encontrado.")
        return pd.DataFrame()

    padrao = re.compile(r"(\d+),\s*(.*?),\s*([\d\.]+),\s*(\d+),\s*([\d\.]+),\s*(\d+),\s*([\d\.]+)")
    dados_extraidos = padrao.findall(conteudo)
    colunas = [
        "Inscrição", "Nome", "Nota_Conhecimentos_Básicos", "Acertos_Básicos",
        "Nota_Conhecimentos_Específicos", "Acertos_Específicos", "Nota_Final"
    ]
    df = pd.DataFrame(dados_extraidos, columns=colunas)
    tipos_de_dados = {
        "Inscrição": int, "Nome": str, "Nota_Conhecimentos_Básicos": str,
        "Acertos_Básicos": int, "Nota_Conhecimentos_Específicos": str,
        "Acertos_Específicos": int, "Nota_Final": str
    }
    
    df = df.astype(tipos_de_dados)
    return df


# --- NOVA FUNÇÃO (MODIFICADA) ---
def salvar_df_como_pdf(df: pd.DataFrame, nome_arquivo_saida: str):
    """
    Salva um DataFrame do pandas em um arquivo PDF com estilo profissional,
    otimizado para caber na página.

    Args:
        df: O DataFrame a ser salvo.
        nome_arquivo_saida: O nome do arquivo PDF a ser gerado (ex: 'relatorio.pdf').
    """
    if df.empty:
        print("DataFrame está vazio. Nenhum PDF será gerado.")
        return

    # Converte o DataFrame para HTML sem o índice da linha
    html_string = df.to_html(index=False, justify='center')

    # CSS para um estilo profissional da tabela
    estilo_css = """
    /* --- ALTERAÇÃO 1: Definir a página como paisagem (landscape) --- */
    @page {
        size: A4 landscape;
        margin: 1.5cm; /* Ajuste de margem se necessário */
    }

    @import url('https://fonts.googleapis.com/css2?family=Poppins:wght@400;600&display=swap');
    
    body {
        font-family: 'Poppins', sans-serif;
        color: #333;
    }
    h1 {
        color: #1a237e;
        text-align: center;
        font-size: 1.5em; /* Reduz um pouco o tamanho do título */
    }
    table {
        width: 100%;
        border-collapse: collapse;
        margin: 20px 0;
        /* --- ALTERAÇÃO 2: Reduzir o tamanho da fonte da tabela --- */
        font-size: 0.75em; 
        box-shadow: 0 0 20px rgba(0, 0, 0, 0.1);
        
        /* --- ALTERAÇÃO 3: Forçar quebra de linha em texto longo --- */
        table-layout: fixed; /* Essencial para o 'word-wrap' funcionar bem */
        word-wrap: break-word;
    }
    th, td {
        /* --- ALTERAÇÃO 4: Reduzir o padding das células --- */
        padding: 8px 10px;
        text-align: left;
    }
    thead tr {
        background-color: #1a237e;
        color: #ffffff;
        text-align: left;
        font-weight: bold;
    }
    tbody tr {
        border-bottom: 1px solid #dddddd;
    }
    tbody tr:nth-of-type(even) {
        background-color: #f3f3f3;
    }
    tbody tr:last-of-type {
        border-bottom: 2px solid #1a237e;
    }
    """

    # Combina o HTML com um título e o CSS
    html_final = f"""
    <html>
      <head>
        <meta charset="UTF-8">
        <title>Relatório de Candidatos</title>
      </head>
      <body>
        <h1>Resultado do Edital</h1>
        {html_string}
      </body>
    </html>
    """

    # Gera o PDF a partir da string HTML e do CSS
    print(f"Gerando PDF '{nome_arquivo_saida}'...")
    css = CSS(string=estilo_css)
    HTML(string=html_final, base_url='.').write_pdf(nome_arquivo_saida, stylesheets=[css])
    print(f"✅ Tabela salva com sucesso em '{nome_arquivo_saida}'!")

# # --- Bloco de Execução Principal ---
if __name__ == "__main__":
    # 1. Simulação: Criando o arquivo 'edital.txt'
    # conteudo_txt = "10088194, Abilio da Cruz Ramos Neto, 32.00, 41, 36.00, 53, 68.00 / 10075745, Abimaeli Simplicio Vasques, 38.00, 41, 48.00, 56, 86.00 / 10136648, Abraao Flavio Nascimento Souza, 21.00, 24, 29.00, 38, 50.00 / 10013311, Abraao Lincoln de Sousa Alves, 24.00, 37, 30.00, 50, 54.00 / 10000737, Abraao Linconies da Silva Campelo, 16.00, 33, 24.00, 47, 40.00 / 10008570, Adaias Abner Brito Silva, 31.00, 37, 44.00, 55, 75.00"
    nome_do_arquivo_entrada = "data/edital_text.txt"
    # with open(nome_do_arquivo_entrada, "w", encoding="utf-8") as f:
    #     f.write(conteudo_txt)

    # 2. Processamento: Cria o DataFrame a partir do arquivo de texto.
    tabela_edital = criar_tabela_do_edital(nome_do_arquivo_entrada)

#     # 3. Exportação: Salva o DataFrame em um arquivo PDF.
#     if not tabela_edital.empty:
#         nome_do_arquivo_saida = "relatorio_edital.pdf"
#         salvar_df_como_pdf(tabela_edital, nome_do_arquivo_saida)


```
