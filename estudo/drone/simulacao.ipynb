```python
from pysimverse import DroneShow

# Inicializa o ambiente de simulação
show = DroneShow()

# Define a quantidade de drones
show.spawn_drones(10)

# Define posições iniciais em linha
for i in range(10):
    show.set_position(i, x=i*2, y=0, z=5)

# Cria uma formação em círculo
show.create_formation("circle", radius=10, center=(0, 0, 10))

# Adiciona efeitos de luz
show.set_light_effect("rainbow")

# Executa a simulação
show.run(duration=30)
```
