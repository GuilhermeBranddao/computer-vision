```python
# Caso Semaforo

# Tempo de verde = 10s
# Tempo de amarelo = 10s
# Tempo de vermelho = 340s

# Gerar numero aleatorios da seguinte forma:
# n <= 30 - verde
# n < n <= 40 - amarelo
# 40 < n <= 340 - Vermelho

import random
import numpy as np

random.seed(42)

N = 1e+05    # 10^5


amostras = np.array([random.randint(1, 340) for _ in range(N)])

# Calculando probabilidades 

num_verde = sum(amostras <= 30)
num_amarelo = sum((amostras > 30) & (amostras <= 40))
num_vermelho = sum(amostras > 40)


num_verde, num_amarelo, num_vermelho
```

```python
import random
import numpy as np

random.seed(42)

N = int(1e5)  # 10^5 como inteiro

amostras = np.array([random.randint(1, 340) for _ in range(N)])

# Calculando probabilidades 
num_verde = sum(amostras <= 30)
num_amarelo = sum((amostras > 30) & (amostras <= 40))
num_vermelho = sum(amostras > 40)

num_verde, num_amarelo, num_vermelho
```

```python
# Quero sabe a probabilidade de dar par em um dado de 6 faces

N = int(1e5)  # 10^5 como inteiro
A = {1, 2, 3, 4, 5, 6} # Espeço amostral

# Gera N amostras aleatórias do conjunto A
D = np.random.choice(list(A), N)

# Verifica se cada elemento é par (2, 4 ou 6)
serve = (D == 2) | (D == 4) | (D == 6)

# Calcula a probabilidade de sair um número par
prob_par = np.sum(serve) / N

prob_par
```

```python
# Quero sabe a probabilidade de dar 7 em dois dados de 6 faces cada

N = int(1e5)  # 10^5 como inteiro
A = {1, 2, 3, 4, 5, 6} # Espeço amostral

# Gera N amostras aleatórias do conjunto A
D1 = np.random.choice(list(A), N)
D2 = np.random.choice(list(A), N)

k = D1 + D2

# Verifica se cada elemento é par (2, 4 ou 6)
serve = (k == 7)

# Calcula a probabilidade de sair um número par
prob_par = np.sum(serve) / N

prob_par
```
