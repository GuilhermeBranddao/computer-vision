```python
import numpy as np
import matplotlib.pyplot as plt

import numpy as np


def distancia_euclidiana(p, q):
    """Calcula a distância Euclidiana entre os pontos p e q."""
    return np.sqrt((p[0] - q[0])**2 + (p[1] - q[1])**2)

def distancia_city_block(p, q):
    """Calcula a distância D4 (City-Block) entre os pontos p e q."""
    return abs(p[0] - q[0]) + abs(p[1] - q[1])

def distancia_chessboard(p, q):
    """Calcula a distância D8 (Chessboard) entre os pontos p e q."""
    return max(abs(p[0] - q[0]), abs(p[1] - q[1]))
```

```python
p = (9, 10)
q = (13, 14)
dist = distancia_euclidiana(p, q)
print(f"A distância Euclidiana entre {p} e {q} é: {dist:.2f}") # Resultado é 5.00



p = (9, 10)
q = (13, 14)
dist = distancia_city_block(p, q)
print(f"A distância City-Block (D4) entre {p} e {q} é: {dist}") # Resultado é 7



p = (9, 10)
q = (13, 14)
dist = distancia_chessboard(p, q)
print(f"A distância Chessboard (D8) entre {p} e {q} é: {dist}") # Resultado é 4
```

```python
plt.scatter(p, q)
plt.plot(p, q, color="red")

dist = distancia_euclidiana(p, q)
# plote um texto no meio da linha entre p e q
plt.text((10 + 13) / 2.29, (10 + 14) / 1.78, f"{dist:.2f}", fontsize=12, ha='center', va='center')

# plt.text((p[0] + q[0]) / 2, (p[1] + q[1]) / 2, f"{dist:.2f}", fontsize=12, ha='center', va='center')
```

```python
# Cria uma imagem 21x21
altura, largura = 21, 21
centro = (10, 10)

# Prepara as imagens de distância
mapa_euclidiano = np.zeros((altura, largura), dtype=np.float32)
mapa_city_block = np.zeros((altura, largura), dtype=np.uint8)
mapa_chessboard = np.zeros((altura, largura), dtype=np.uint8)

# Calcula a distância de cada pixel ao centro para cada métrica
for y in range(altura):
    for x in range(largura):
        ponto_atual = (x, y)
        mapa_euclidiano[y, x] = distancia_euclidiana(centro, ponto_atual)
        mapa_city_block[y, x] = distancia_city_block(centro, ponto_atual)
        mapa_chessboard[y, x] = distancia_chessboard(centro, ponto_atual)

# Exibe os resultados
plt.figure(figsize=(18, 5))

plt.subplot(1, 3, 1)
plt.imshow(mapa_euclidiano, cmap='viridis')
plt.title('Distância Euclidiana ($D_e$) - Forma de Círculo')
plt.colorbar()

plt.subplot(1, 3, 2)
plt.imshow(mapa_city_block, cmap='viridis')
plt.title('Distância City-Block ($D_4$) - Forma de Losango')
plt.colorbar()

plt.subplot(1, 3, 3)
plt.imshow(mapa_chessboard, cmap='viridis')
plt.title('Distância Chessboard ($D_8$) - Forma de Quadrado')
plt.colorbar()

plt.show()
```
