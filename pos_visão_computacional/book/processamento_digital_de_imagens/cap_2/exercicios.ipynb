Utilizando as informações apresentadas na Seção 2.1 e pensando puramente em termos geométricos, 
estime o diâmetro do menor ponto impresso que o olho é capaz de discernir se a página na qual o ponto foi impresso estiver a 0,2 metros de distância dos olhos. 

Suponha, para fins de simplificação, que o sistema visual deixe de detectar o ponto quando a imagem do ponto na fóvea ficar menor do que o diâmetro de um receptor (cone) 
nessa área da retina. 

Suponha também que a fóvea possa ser modelada em um arranjo matricial quadrado de dimensões 1,5 mm × 1,5 mm e que os cones e es paços entre eles sejam distribuídos uniformemente ao 
longo desse arranjo.





````python
import numpy as np

# --- Passo 1: Definir as constantes e informações do problema ---
# Distâncias em milímetros (mm)
distancia_objeto_mm = 200.0  # 0,2 metros
distancia_focal_mm = 17.0     # Modelo simplificado do olho

# Parâmetros da fóvea
lado_fovea_mm = 1.5
num_cones = 337_000 # Usando '_' para facilitar a leitura de números grandes

# --- Passo 2: Calcular o diâmetro de um único cone ---
area_fovea_mm2 = lado_fovea_mm ** 2
area_por_cone_mm2 = area_fovea_mm2 / num_cones
diametro_cone_mm = np.sqrt(area_por_cone_mm2)

# --- Passo 3: Aplicar a fórmula de triângulos semelhantes ---
# d_ponto / D_objeto = d_cone / d_focal  =>  d_ponto = D_objeto * (d_cone / d_focal)
diametro_ponto_mm = distancia_objeto_mm * (diametro_cone_mm / distancia_focal_mm)


# --- Passo 4: Análise e Conclusão ---
print("--- Análise do Exercício 2.1 ---")
print(f"Área da fóvea: {area_fovea_mm2:.2f} mm²")
print(f"Número de cones considerado: {num_cones}")
print(f"Diâmetro médio de um cone: {diametro_cone_mm * 1000:.2f} micrômetros")
print("\n--- Resultado ---")
print(f"O diâmetro do menor ponto discernível é de aproximadamente {diametro_ponto_mm:.4f} mm.")

# --- Bônus: O que isso significa em termos de impressão (DPI)? ---
# DPI = Dots Per Inch (Pontos Por Polegada)
# 1 polegada = 25.4 mm
diametro_ponto_polegadas = diametro_ponto_mm / 25.4
dpi_equivalente = 1 / diametro_ponto_polegadas

print("\n--- Análise Prática (Impressão) ---")
print(f"Isso equivale a uma resolução de impressão de aproximadamente {dpi_equivalente:.0f} DPI.")
````

Releia a discussão no final da Seção 2.5.2, na qual definimos o fundo como (R_u)^c, que é o complemento da união de todas as regiões de uma imagem.

### **Aula: Redefinindo o "Fundo" para Excluir Buracos**

#### **Passo 1: Entendendo o Problema com uma Analogia Visual**

Primeiro, vamos visualizar o que o exercício descreve.

  * `Ru`: É a união de todas as regiões de interesse (o *foreground*). Em nossa analogia do mapa, `Ru` são todos os **países** juntos.
  * `(Ru)c`: É o complemento, ou seja, tudo o que não é `Ru`. Em nosso mapa, isso seria toda a **água** do mundo, o que inclui tanto o **oceano principal** quanto os **lagos isolados** dentro dos países.

O problema é que, pela definição padrão, esses lagos isolados são considerados "fundo". O exercício quer que a gente encontre uma forma de definir o fundo como sendo *apenas o oceano principal*, excluindo os lagos.

**Diagrama do Problema:**

```
B B B B B B B B B B B B
B R R R R R R R R R R B   <- R é a Região (país)
B R R R R R R R R R R B
B R R  H H H  R R R R B   <- H é o Buraco (lago)
B R R  H H H  R R R R B
B R R R R R R R R R R B
B R R R R R R R R R R B
B B B B B B B B B B B B   <- B é o Fundo "Verdadeiro" (oceano)
```

Na definição padrão, Fundo = `B + H`.
Na definição que queremos, Fundo = `B`.

A proibição de dizer "o fundo é o fundo menos os buracos" significa que precisamos de uma definição *construtiva*. Precisamos de uma propriedade única que o "Fundo Verdadeiro" (`B`) tenha e que os "Buracos" (`H`) não tenham.

#### **Passo 2: Usando a Dica - A Magia da Conectividade**

A dica é a chave de tudo: **conectividade**.

Vamos pensar nos conjuntos de pixels de fundo `B` e `H` do nosso diagrama.

  * O conjunto `B` é um **componente conexo**. Todos os pixels em `B` estão conectados entre si.
  * O conjunto `H` também é um **componente conexo**. Todos os pixels em `H` estão conectados entre si.

Então, tanto o oceano quanto o lago são, individualmente, componentes conexos. O que os diferencia?

Pense no mapa. Qual a diferença fundamental entre o oceano e um lago?
**O oceano toca a borda do mapa. O lago não.**

Esta é a propriedade que procurávamos\! Em termos de processamento de imagens:

> O "Fundo Verdadeiro" é o **único** componente conexo do fundo inicial que está **conectado à fronteira da imagem**. Qualquer outro componente conexo do fundo que *não* toca a fronteira da imagem é, por definição, um "buraco".

#### **Passo 3: Formulando a Nova Definição (A Resposta Formal)**

Agora podemos construir nossa resposta sem ser circular.

Seja `S_bg = (Ru)c` o conjunto de todos os pixels de fundo iniciais. A nova definição de fundo, que chamaremos de `Fundo_Verdadeiro`, pode ser definida da seguinte forma:

> O **Fundo\_Verdadeiro** é o conjunto de todos os pixels `p` que pertencem a `S_bg` e para os quais existe um **caminho** do pixel `p` até qualquer pixel `q` que esteja na **fronteira da imagem**, sendo que este caminho é formado inteiramente por pixels que também pertencem a `S_bg`.

De forma mais algorítmica:

1.  Defina o conjunto de fundo inicial `S_bg = (Ru)c`.
2.  Encontre todos os componentes conexos dentro de `S_bg`.
3.  Identifique qual desses componentes conexos contém pixels da fronteira da imagem (por exemplo, pixels da linha 0, coluna 0, etc.).
4.  Este componente conexo específico é o novo fundo. Os demais são os buracos.


````python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# --- 1. Criar uma imagem com uma região (R) e um buraco (H) ---
imagem = np.zeros((200, 300), dtype=np.uint8)
# Desenha um grande retângulo branco (nossa região R)
cv2.rectangle(imagem, (30, 30), (270, 170), 255, -1)
# Desenha um pequeno retângulo preto no meio (nosso buraco H)
cv2.rectangle(imagem, (100, 80), (200, 120), 0, -1)


# --- 2. Encontrar o fundo inicial (Ru)^c ---
# No nosso caso, (Ru)^c são todos os pixels com valor 0.
# O OpenCV trabalha melhor com o objeto em branco e o fundo em preto.
# Então, vamos inverter a imagem para encontrar os componentes do fundo.
fundo_inicial = cv2.bitwise_not(imagem)


# --- 3. Encontrar TODOS os componentes conexos do fundo inicial ---
# Esta função vai rotular o "oceano" (B) com um número (ex: 1)
# e o "lago" (H) com outro número (ex: 2).
num_labels, labels_im = cv2.connectedComponents(fundo_inicial)

# Para visualizar, vamos colorir os componentes
def colorir_componentes(labels):
    """Cria uma imagem colorida para cada componente conexo."""
    label_hue = np.uint8(179 * labels / np.max(labels))
    blank_ch = 255 * np.ones_like(label_hue)
    labeled_img = cv2.merge([label_hue, blank_ch, blank_ch])
    labeled_img = cv2.cvtColor(labeled_img, cv2.COLOR_HSV2BGR)
    labeled_img[label_hue == 0] = 0 # Pinta o fundo (rótulo 0) de preto
    return labeled_img

componentes_fundo_coloridos = colorir_componentes(labels_im)


# --- 4. Identificar o componente do "Fundo Verdadeiro" ---
# Sabemos que o fundo verdadeiro toca a borda da imagem.
# Vamos pegar o rótulo do pixel no canto superior esquerdo (0,0).
fundo_verdadeiro_label = labels_im[0, 0]


# --- 5. Criar a máscara final do fundo, excluindo os buracos ---
fundo_final = np.zeros_like(imagem)
# Mantém apenas os pixels que têm o mesmo rótulo do pixel (0,0)
fundo_final[labels_im == fundo_verdadeiro_label] = 255


# --- Exibindo os resultados do nosso algoritmo ---
plt.figure(figsize=(12, 8))
plt.subplot(2, 2, 1)
plt.imshow(imagem, cmap='gray')
plt.title('1. Imagem Original com Região (R) e Buraco (H)')

plt.subplot(2, 2, 2)
plt.imshow(fundo_inicial, cmap='gray')
plt.title('2. Fundo Inicial (Ru)^c = B + H')

plt.subplot(2, 2, 3)
plt.imshow(componentes_fundo_coloridos)
plt.title('3. Componentes do Fundo (B e H com rótulos diferentes)')

plt.subplot(2, 2, 4)
plt.imshow(fundo_final, cmap='gray')
plt.title('4. Fundo Final (Apenas B)')

plt.tight_layout()
plt.show()
````
