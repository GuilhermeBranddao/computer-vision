```python
from PIL.ImageChops import difference
from PIL import Image

original = Image.open("data\images\images_com_diferenças\imagens_7_erros\original.png")
modificada = Image.open("data\images\images_com_diferenças\imagens_7_erros\modificada.png")

original = original.convert("RGB")
modificada = modificada.convert("RGB")

modificada = modificada.resize((original.width, original.height), Image.Resampling.LANCZOS)

difference(original, modificada).show()
```

```python
from PIL import Image, ImageChops

original = Image.open("data/images/images_com_diferenças/imagens_7_erros/original.png").convert("RGB")
modificada = Image.open("data/images/images_com_diferenças/imagens_7_erros/modificada.png").convert("RGB")

modificada = modificada.resize((original.width, original.height), Image.Resampling.LANCZOS)

diff = ImageChops.difference(original, modificada)
diff = diff.convert("L")  # Escala de cinza
diff = diff.point(lambda x: x * 10)  # Amplifica contraste
diff.show()
```
