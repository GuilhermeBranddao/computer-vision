```python
from skimage.io import imread
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # necessário para projeção 3D
from matplotlib.ticker import LinearLocator, FormatStrFormatter

def plot_3d(x, y, z, cmap="Reds", title=""):
    """
    Plota uma visualização 3D de um canal de imagem.

    x, y, z devem ser arrays numpy compatíveis com a função plot_surface.
    """

    fig = plt.figure(figsize=(15, 15))
    ax = fig.add_subplot(111, projection='3d')

    surf = ax.plot_surface(X=x, Y=y, Z=z, cmap=cmap, linewidth=0,
                           antialiased=False, rstride=2, cstride=2, alpha=0.5)

    ax.xaxis.set_major_locator(LinearLocator(10))
    ax.xaxis.set_major_formatter(FormatStrFormatter('%.02f'))
    ax.view_init(elev=10., azim=5)
    ax.set_title(title, size=20)
    plt.show()
```
