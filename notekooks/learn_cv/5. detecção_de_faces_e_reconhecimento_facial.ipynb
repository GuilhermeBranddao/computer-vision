- Caracteristicas
    - PCA é um algoritmo não supervisionado 
    - PCA é uma operação matematica que reduz a dimencionalidade dados e enconra os componentes proncipais 
        - É reduzida a 3/4/5 componentes, os componentes, cada um deles expressa a relação entre os recursos/variaveis que estiverem em meu conjunto de dados 
        - Ou seja o PCA recebe um varias variaveis e converte a 3/4/5 componentes e cada um deles representa cada uma das variaveis que eu tinha no inicio
