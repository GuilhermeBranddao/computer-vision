```python
from PIL import ImageEnhance
from pathlib import Path

def open_image(path:Path):
    return Image.open(path)

def save_image(img:Image, path:Path):
    img.save(path)

def crop_image(img:Image, dim):
    return img.crop(dim)

def show_image(img:Image):
    img.show()

def convert_to_grayscale(img:Image):
    return img.convert("L")

def rotate_image(img:Image, angle, expand:bool=True):
    return img.rotate(angle, expand=expand)

def enhance_brightness(img:Image, factor:int):
    enhancer = ImageEnhance.Brightness(img)
    return enhancer.enhance(factor)

def enhance_contrast(img:Image, factor:int):
    enhancer = ImageEnhance.Contrast(img)
    return enhancer.enhance(factor)

def get_pixel_value(img:Image, coord:tuple[int, int]):
    return img.getpixel(coord)
  

def set_pixel_value(img:Image, coord:tuple[int, int], value:tuple[int, int, int]):
    img.putpixel(coord, value)
    return img

def get_pixel_values(img:Image, coords:list[tuple[int, int]]):
    return [img.getpixel(coord) for coord in coords]

def get_image_size(img:Image):
    return img.size

def get_image_format(img:Image):
    return img.format

def get_image_mode(img:Image):
    return img.mode

def get_image_info(img:Image):
    return img.info

def get_image_palette(img:Image):
    if img.mode == "P":
        return img.getpalette()
    else:
        raise ValueError("Image does not have a palette")
```
