```python
import pandas as pd

# Parâmetros iniciais
meses_economizando = 25
salario_inicial = 7326
percentual_economia = 0.5
reajuste_anual = 0.00  # 5% ao ano é 0.05
rendimento_mensal = 0.0093  # cerca de 100% do CDI (~11.7% a.a.)
dolar_hoje = 5.60

# Variáveis de controle
na_caixinha = 0
na_caixinha_s_rendimento = 0
lista_infos = []

# Simulação
for mes in range(1, meses_economizando + 1):
    anos_passados = (mes - 1) // 12
    salario_atual = salario_inicial * (1 + reajuste_anual) ** anos_passados
    economia_mes = salario_atual * percentual_economia
    na_caixinha += economia_mes
    na_caixinha_s_rendimento += economia_mes
    # Aplica rendimento sobre o total acumulado
    na_caixinha *= (1 + rendimento_mensal)

    lista_infos.append({
        "Mês": mes,
        "Salário": round(salario_atual, 2),
        "Economia_Mensal": round(economia_mes, 2),
        "Patrimônio c/ Rendimento (R$)": round(na_caixinha, 2),
        "Patrimônio s/ Rendimento (R$)": na_caixinha_s_rendimento,
        "Patrimônio (US$)": round(na_caixinha / dolar_hoje, 2),
    })

df = pd.DataFrame(lista_infos)
df
```
